Now Let's Create Our First Moving Average

```python
# Create 5-day Moving Average
stock_data["MA_5"] = stock_data["Close"].rolling(window=5).mean()

# Display first 10 rows
stock_data.head(10)
```

```python
stock_data["MA_5"] = stock_data["Close"].rolling(window=5).mean()
```

Create the 5-Day and 10-Day Moving Averages

```python
# Create Moving Averages
stock_data["MA_5"] = stock_data["Close"].rolling(window=5).mean()
stock_data["MA_10"] = stock_data["Close"].rolling(window=10).mean()

# Display the first 15 rows
stock_data.head(15)
```

```python
# Create Daily Return feature
stock_data["Daily_Return"] = stock_data["Close"].pct_change()

# Display first 10 rows
stock_data.head(10)
```

```python
# Create Daily Return feature
stock_data["Daily_Return"] = stock_data["Close"].pct_change()

# Display the first 10 rows
stock_data.head(10)
```

We calculate the standard deviation of the recent daily returns.
Don't worry about the mathematics yet.
Just remember:
Standard deviation tells us how spread out the values are.
Small standard deviation = Stable prices
Large standard deviation = Unstable prices

```python
# Create 5-day Rolling Volatility
stock_data["Volatility_5"] = stock_data["Daily_Return"].rolling(window=5).std()

# Display first 15 rows
stock_data.head(15)
```

```python
# Create Target Variable (Tomorrow's Close)
stock_data["Target"] = stock_data["Close"].shift(-1)

stock_data.head(10)
```
