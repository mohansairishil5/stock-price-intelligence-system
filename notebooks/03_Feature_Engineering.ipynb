Create EMA (Exponential Moving Average)
Before writing any code, let's understand why we use it
What is EMA?
EMA is similar to a moving average, but it gives more weight to recent prices. That means it reacts faster to changes in the market than a Simple Moving Average (SMA).
SMA: Every value in the window has equal importance.
EMA: Recent prices have more influence.
This makes EMA useful for detecting trend changes earlier.

```python
# Create Exponential Moving Averages
stock_data["EMA_12"] = stock_data["Close"].ewm(span=12, adjust=False).mean()

stock_data["EMA_26"] = stock_data["Close"].ewm(span=26, adjust=False).mean()

# View the new columns
stock_data[["Close", "EMA_12", "EMA_26"]].head(15)
```

RSI (Relative Strength Index)
This is one of the most important technical indicators you'll learn.
What is RSI?
RSI measures the speed and strength of recent price movements. It tells us whether a stock may be:
Overbought (price has risen a lot recently)
Oversold (price has fallen a lot recently)
RSI Range
RSI Value	Meaning
0–30	Oversold (possible buying opportunity)
30–70	Neutral
70–100	Overbought (possible selling opportunity)
Remember, these are signals, not guarantees. Traders usually combine RSI with other indicators.

Calculate the daily price change

```python
delta = stock_data["Close"].diff()
```

Separate gains and losses

```python
gain = delta.where(delta > 0, 0)
loss = -delta.where(delta < 0, 0)
```

: Calculate the 14-day average gain and loss

```python
avg_gain = gain.rolling(window=14).mean()
avg_loss = loss.rolling(window=14).mean()
```

Compute RS (Relative Strength)

```python
rs = avg_gain / avg_loss
```

Compute RSI

```python
stock_data["RSI_14"] = 100 - (100 / (1 + rs))
```

```python
# Calculate price change
delta = stock_data["Close"].diff()

# Separate gains and losses
gain = delta.where(delta > 0, 0)
loss = -delta.where(delta < 0, 0)

# Calculate average gain and average loss
avg_gain = gain.rolling(window=14).mean()
avg_loss = loss.rolling(window=14).mean()

# Calculate Relative Strength
rs = avg_gain / avg_loss

# Calculate RSI
stock_data["RSI_14"] = 100 - (100 / (1 + rs))

# Display results
stock_data[["Close", "RSI_14"]].head(20)
```

```python
stock_data[["Close", "RSI_14"]].head(20)
```

INDICATOR-2
MACD stands for Moving Average Convergence Divergence.
MACD = EMA_12 - EMA_26
**MACD Line
Signal Line (9-day EMA of the MACD line)
Histogram (difference between MACD and Signal Line)**

```python
# Calculate MACD Line
stock_data["MACD"] = stock_data["EMA_12"] - stock_data["EMA_26"]

# Calculate Signal Line (9-day EMA of MACD)
stock_data["Signal_Line"] = stock_data["MACD"].ewm(span=9, adjust=False).mean()

# Calculate Histogram
stock_data["MACD_Histogram"] = stock_data["MACD"] - stock_data["Signal_Line"]

# View the results
stock_data[
    ["Close", "EMA_12", "EMA_26", "MACD", "Signal_Line", "MACD_Histogram"]
].head(20)
```

Bollinger Bands-4

This is another widely used indicator.

You'll create three new features:

BB_Middle (20-day moving average)
BB_Upper
BB_Lower

These bands help capture volatility and show when prices are unusually far from their recent average.

```python
# Calculate 20-day Moving Average
stock_data["BB_Middle"] = stock_data["Close"].rolling(window=20).mean()

# Calculate 20-day Standard Deviation
std_20 = stock_data["Close"].rolling(window=20).std()

# Calculate Upper and Lower Bands
stock_data["BB_Upper"] = stock_data["BB_Middle"] + (2 * std_20)
stock_data["BB_Lower"] = stock_data["BB_Middle"] - (2 * std_20)

# View results
stock_data[
    ["Close", "BB_Middle", "BB_Upper", "BB_Lower"]
].head(25)
```

ATR (Average True Range)
ATR measures how much the stock typically moves in a day, regardless of whether it moves up or down.
Unlike Bollinger Bands, which look at volatility over a rolling window, ATR focuses on the daily trading range

true Range (TR)

```python
high_low = stock_data["High"] - stock_data["Low"]

high_close = (stock_data["High"] - stock_data["Close"].shift()).abs()

low_close = (stock_data["Low"] - stock_data["Close"].shift()).abs()
```

True Range:Take the maximum of those three values:

```python
stock_data["TR"] = pd.concat(
    [high_low, high_close, low_close],
    axis=1
).max(axis=1)
```

ATR (14-day average)

```python
# Step 1: Calculate the three True Range components
high_low = stock_data["High"] - stock_data["Low"]

high_close = (stock_data["High"] - stock_data["Close"].shift()).abs()

low_close = (stock_data["Low"] - stock_data["Close"].shift()).abs()

# Step 2: Take the maximum of the three
stock_data["TR"] = pd.concat(
    [high_low, high_close, low_close],
    axis=1
).max(axis=1)

# Step 3: Calculate 14-day ATR
stock_data["ATR_14"] = stock_data["TR"].rolling(window=14).mean()

# Step 4: Display the result
stock_data[
    ["High", "Low", "Close", "TR", "ATR_14"]
].head(20)
```

```python
stock_data[
    ["High", "Low", "Close", "TR", "ATR_14"]
].head(20)
```

Momentum
What is Momentum?
Momentum answers a simple question:
"How much has the stock price changed compared to N days ago?
Formula:
For a 10-day momentum:
Momentum = Today's Close − Close 10 days ago

```python
# Calculate 10-day Momentum
stock_data["Momentum_10"] = stock_data["Close"] - stock_data["Close"].shift(10)

# View results
stock_data[["Close", "Momentum_10"]].head(20)
```

ROC (Rate of Change)
Momentum tells us the difference in price.
ROC tells us the percentage change.
This makes it easier to compare stocks with different price ranges.
Formula:
ROC = ((Today's Close − Close 10 days ago)
       / Close 10 days ago) × 100

```python
# Calculate 10-day Rate of Change
stock_data["ROC_10"] = (
    (stock_data["Close"] - stock_data["Close"].shift(10))
    / stock_data["Close"].shift(10)
) * 100

# View results
stock_data[["Close", "Momentum_10", "ROC_10"]].head(20)
```
