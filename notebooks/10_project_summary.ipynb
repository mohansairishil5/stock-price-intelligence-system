1.3 Problem Statement

# Problem Statement
Stock prices change continuously due to market activity, making accurate prediction a challenging task.
The goal of this project is to learn patterns from historical stock prices and engineered features to estimate the next trading day's closing price.

Let's analyze the problem statement
Notice we didn't write:
"This project predicts the future perfectly."
That would be incorrect.
Instead, we wrote:
"Learn patterns."
Machine learning doesn't know the future. It identifies relationships in historical data and uses them to estimate future values.

Why is stock prediction difficult?
Think about what can affect a stock price:
Company earnings
Interest rates
Inflation
News
Product launches
Competitor performance
Global events
Investor sentiment
Our model doesn't know any of these
It only knows what is present in the dataset.
That's why we say estimate, not guarantee.
