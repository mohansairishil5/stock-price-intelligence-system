```python
ticker = "AAPL"

stock_data = yf.download(
    ticker,
    start="2020-01-01",
    end="2025-12-31"
)

stock_data.head()
```

## Exploratory Data Analysis (EDA)

```python
# Plot the closing price over time

plt.figure(figsize=(15,6))

plt.plot(stock_data.index,
         stock_data[("Close", "AAPL")],
         label="Closing Price")

plt.title("Apple Closing Stock Price (2020-2025)")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend()

plt.show()
```

Visualization 2: Trading Volume
Business Question

Was the stock traded equally every day?

```python
plt.figure(figsize=(15,6))

plt.plot(
    stock_data.index,
    stock_data[("Volume","AAPL")]
)

plt.title("Apple Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")

plt.show()
```

📊 Visualization 3: Daily High vs Daily Low
Question:How much does the stock fluctuate within a day?

```python
plt.figure(figsize=(15,6))

plt.plot(
    stock_data.index,
    stock_data[("High","AAPL")],
    label="High"
)

plt.plot(
    stock_data.index,
    stock_data[("Low","AAPL")],
    label="Low"
)

plt.title("Daily High and Low Prices")
plt.xlabel("Date")
plt.ylabel("Price")

plt.legend()

plt.show()
```

📊 Visualization 4: Histogram of Closing Price
Business Question:
At what price range did Apple spend most of its time?

```python
plt.figure(figsize=(10,6))

plt.hist(
    stock_data[("Close","AAPL")],
    bins=30
)

plt.title("Distribution of Closing Price")
plt.xlabel("Closing Price")
plt.ylabel("Frequency")

plt.show()
```
