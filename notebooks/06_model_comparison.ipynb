```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
comparison_df = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "XGBoost"
    ],
    "MAE": [
        3.8766,
        13.6158,
        14.5774
    ],
    "RMSE": [
        5.4467,
        20.0967,
        21.5460
    ],
    "R2 Score": [
        0.9509,
        0.3318,
        0.2319
    ]
})

comparison_df
```

```python
best_model = comparison_df.loc[
    comparison_df["R2 Score"].idxmax()
]

print(best_model)
```

```python
print("Best Model:", best_model["Model"])
```

```python
plt.figure(figsize=(8,5))

plt.bar(comparison_df["Model"], comparison_df["MAE"])

plt.title("Model Comparison - MAE")
plt.xlabel("Models")
plt.ylabel("MAE")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(comparison_df["Model"], comparison_df["RMSE"])

plt.title("Model Comparison - RMSE")
plt.xlabel("Models")
plt.ylabel("RMSE")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(comparison_df["Model"], comparison_df["R2 Score"])

plt.title("Model Comparison - R² Score")
plt.xlabel("Models")
plt.ylabel("R² Score")

plt.show()
```

```python
print("=" * 50)
print("🏆 FINAL MODEL SELECTION")
print("=" * 50)

print(f"Selected Model : {best_model['Model']}")
print(f"MAE            : {best_model['MAE']:.4f}")
print(f"RMSE           : {best_model['RMSE']:.4f}")
print(f"R² Score       : {best_model['R2 Score']:.4f}")
```

# Conclusion

Three machine learning models were evaluated for Apple stock price prediction:

- Linear Regression
- Random Forest
- XGBoost

The models were compared using MAE, RMSE, and R² Score.

## Results

| Model | MAE | RMSE | R² Score |
|--------|-----|------|----------|
| Linear Regression | 3.8766 | 5.4467 | 0.9509 |
| Random Forest | 13.6158 | 20.0967 | 0.3318 |
| XGBoost | 14.5774 | 21.5460 | 0.2319 |

## Final Selection

Linear Regression achieved the lowest prediction error and the highest R² score. Therefore, it was selected as the final model for this project.
